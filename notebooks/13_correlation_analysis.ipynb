```python
import pandas as pd
import numpy as np

from scipy.stats import pearsonr
```

```python
topology_path = "resnet_epoch1_layerwise_topology.csv"

df = pd.read_csv(topology_path)

print(df)
```

```python
train_accuracy = 25.36
test_accuracy = 23.50

generalization_gap = (
    train_accuracy - test_accuracy
)

print("Training accuracy:", train_accuracy)
print("Test accuracy:", test_accuracy)
print("Generalization gap:", generalization_gap)
```

```python
num_training_states = 1

print(
    "Number of training-state observations:",
    num_training_states
)

if num_training_states < 2:
    print(
        "\nPaper-style topology-accuracy correlation "
        "cannot be estimated from the current experiment."
    )
```

```python
paper_style_correlation = np.nan

print(
    "Topology-accuracy Pearson r:",
    paper_style_correlation
)

print(
    "Reason: only one training checkpoint is available."
)
```

```python
depth = df["Relative_Depth"].values

h0 = df["H0_Descriptor"].values
h1 = df["H1_Descriptor"].values
phdim = df["PHdim"].values

r_h0_depth, p_h0_depth = pearsonr(
    depth,
    h0
)

r_h1_depth, p_h1_depth = pearsonr(
    depth,
    h1
)

r_phdim_depth, p_phdim_depth = pearsonr(
    depth,
    phdim
)

print("H0 vs depth:")
print("r =", r_h0_depth)
print("p =", p_h0_depth)

print("\nH1 vs depth:")
print("r =", r_h1_depth)
print("p =", p_h1_depth)

print("\nPHdim vs depth:")
print("r =", r_phdim_depth)
print("p =", p_phdim_depth)
```

```python
correlation_df = df[
    [
        "Relative_Depth",
        "H0_Descriptor",
        "H1_Descriptor",
        "PHdim"
    ]
].corr()

print(
    correlation_df
)
```

```python
accuracy_summary = pd.DataFrame({
    "Training_State": ["Epoch 1"],
    "Train_Accuracy": [train_accuracy],
    "Test_Accuracy": [test_accuracy],
    "Generalization_Gap": [generalization_gap]
})

print(
    accuracy_summary.to_string(index=False)
)
```

```python
correlation_summary = pd.DataFrame({
    "Measure": [
        "H0 vs Relative Depth",
        "H1 vs Relative Depth",
        "PHdim vs Relative Depth",
        "Topology vs Test Accuracy"
    ],
    "Pearson_r": [
        r_h0_depth,
        r_h1_depth,
        r_phdim_depth,
        np.nan
    ],
    "Interpretation": [
        "Layer-wise relationship",
        "Layer-wise relationship",
        "Layer-wise relationship",
        "Not estimable: one checkpoint"
    ]
})

print(
    correlation_summary.to_string(index=False)
)
```

```python
correlation_summary.to_csv(
    "resnet_epoch1_correlation_summary.csv",
    index=False
)

print(
    "Saved: "
    "resnet_epoch1_correlation_summary.csv"
)
```
