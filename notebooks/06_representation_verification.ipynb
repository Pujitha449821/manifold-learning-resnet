```python
import torch

vectorized_block_features = torch.load(
    "./resnet_epoch1_vectorized_representations.pt",
    map_location="cpu"
)

all_labels = torch.load(
    "./resnet_epoch1_labels.pt",
    map_location="cpu"
)

print("Number of blocks:", len(vectorized_block_features))
print("Number of labels:", len(all_labels))
```

```python
for block_idx, vectors in enumerate(
    vectorized_block_features
):

    print(
        f"Block {block_idx + 1}:",
        vectors.shape
    )
```

```python
for block_idx, vectors in enumerate(
    vectorized_block_features
):

    has_nan = torch.isnan(vectors).any().item()
    has_inf = torch.isinf(vectors).any().item()

    print(
        f"Block {block_idx + 1}: "
        f"NaN={has_nan}, "
        f"Inf={has_inf}"
    )
```

```python
for block_idx, vectors in enumerate(
    vectorized_block_features
):

    print(f"\nBlock {block_idx + 1}")

    print(
        "Minimum:",
        vectors.min().item()
    )

    print(
        "Maximum:",
        vectors.max().item()
    )

    print(
        "Mean:",
        vectors.mean().item()
    )

    print(
        "Std:",
        vectors.std().item()
    )
```

```python
for block_idx, vectors in enumerate(
    vectorized_block_features
):

    feature_std = vectors.std(
        dim=0
    ).mean().item()

    print(
        f"Block {block_idx + 1}: "
        f"average feature std = {feature_std:.6f}"
    )
```

```python
for block_idx in range(9):

    current = vectorized_block_features[
        block_idx
    ]

    next_block = vectorized_block_features[
        block_idx + 1
    ]

    difference = torch.mean(
        torch.abs(
            next_block - current
        )
    ).item()

    print(
        f"Block {block_idx + 1} → "
        f"Block {block_idx + 2}: "
        f"mean absolute difference = "
        f"{difference:.6f}"
    )
```

```python
print("Number of labels:", len(all_labels))

print(
    "First 20 labels:",
    all_labels[:20].tolist()
)

print(
    "Unique labels:",
    torch.unique(all_labels).tolist()
)
```

```python
unique, counts = torch.unique(
    all_labels,
    return_counts=True
)

for label, count in zip(unique, counts):

    print(
        f"Class {label.item()}: "
        f"{count.item()} samples"
    )
```
